```python
import pandas as pd
import matplotlib.pyplot as plt

# Using the converted absolute values in Millions of TJ
years = list(range(2000, 2024))
volumes = [4.70, 4.65, 3.99, 4.77, 5.27, 5.07, 4.94, 4.82, 4.42, 4.62, 5.23, 4.99, 5.06, 4.47, 4.54, 4.52, 3.78, 3.81, 3.99, 4.22, 3.23, 2.70, 2.40, 2.66]

plt.figure(figsize=(10, 5))
plt.fill_between(years, volumes, color="#2ca02c", alpha=0.4)
plt.plot(years, volumes, color="#2ca02c", linewidth=2)

plt.title("Total Magnitude of Nigeria's Crude Oil Exports (2000–2023)", fontsize=14, fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Export Volume (Million TJ)")
plt.xlim(2000, 2023)
plt.ylim(0, 6)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Year': list(range(2000, 2024)),
    'Volume': [4.70, 4.65, 3.99, 4.77, 5.27, 5.07, 4.94, 4.82, 4.42, 4.62, 5.23, 4.99, 5.06, 4.47, 4.54, 4.52, 3.78, 3.81, 3.99, 4.22, 3.23, 2.70, 2.40, 2.66]
}
df = pd.DataFrame(data)

plt.figure(figsize=(10, 5))
# sns.regplot automatically plots the data points and calculates a line of best fit
sns.regplot(x='Year', y='Volume', data=df, color='#d62728', marker='x', scatter_kws={'s':50})

plt.title("Scatter Plot with Linear Trendline of Crude Oil Exports", fontsize=14, fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Export Volume (Million TJ)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
