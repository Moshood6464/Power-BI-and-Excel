```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Recreate the dataset from the image
data = {
    'Year': list(range(2000, 2024)),
    'Crude_Oil_TJ': [
        4923513, 5160417, 4429358, 5041538, 5449430, 5499316, 5228556, 4880595,
        4668570, 4678701, 5407965, 5239072, 5184954, 4693492, 4691141, 4535371,
        3917677, 3979130, 4084111, 4295487, 3307055, 2944359, 2555025, 2737203
    ]
}

df = pd.DataFrame(data)
```

```python

# 2. Basic Descriptive Statistics
print("--- Summary Statistics ---")
print(df.describe())
```

```python
# 3. Feature Engineering: Year-over-Year (YoY) Percentage Change
df['YoY_Growth_%'] = df['Crude_Oil_TJ'].pct_change() * 100
```

```python
# Finding key milestones
peak_year = df.loc[df['Crude_Oil_TJ'].idxmax()]
lowest_year = df.loc[df['Crude_Oil_TJ'].idxmin()]

print(f"\nPeak Production Year: {int(peak_year['Year'])} ({peak_year['Crude_Oil_TJ']:,} TJ)")
print(f"Lowest Production Year: {int(lowest_year['Year'])} ({lowest_year['Crude_Oil_TJ']:,} TJ)")
```

```python
# 4. Data Visualization
plt.figure(figsize=(14, 6))
```

```python
import matplotlib.pyplot as plt
import numpy as np

# -------------------------------------------------------------------------
# 1. DATA PREPARATION (Estimated based on the visual trends)
# -------------------------------------------------------------------------
years = np.arange(2000, 2024)

# Mock data in Terajoules (TJ) reflecting the trends in the screenshots
production_data = [
    4.5e6, 4.7e6, 4.3e6, 4.9e6, 5.1e6, 5.2e6, 4.9e6, 4.7e6, 4.5e6, 4.5e6,
    5.1e6, 4.9e6, 4.8e6, 4.5e6, 4.5e6, 4.1e6, 4.2e6, 4.4e6, 3.6e6, 3.3e6,
    3.0e6, 2.7e6, 2.9e6, 3.1e6
]

# Exports closely track production but are slightly lower
exports_data = [val * 0.95 for val in production_data]

# Set a clean, minimalist style
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['axes.edgecolor'] = '#CCCCCC'
plt.rcParams['axes.linewidth'] = 0.8

# -------------------------------------------------------------------------
# CHART 1: Evolution of crude oil production, Nigeria (Area Chart)
# -------------------------------------------------------------------------
fig1, ax1 = plt.subplots(figsize=(6, 5), facecolor='#F9F9F9')
ax1.set_facecolor('#F9F9F9')

# Plot the area chart
ax1.fill_between(years, production_data, color='#FFC466', alpha=0.9)
ax1.plot(years, production_data, color='#E69526', linewidth=2)

# Grid and styling
ax1.axhline(y=6000000, color='#999999', linestyle='--', linewidth=0.8)
ax1.text(2023, 6100000, '6 000 000\nTJ', ha='right', va='bottom', color='#555555', fontsize=9)

# Axis formatting
ax1.set_xlim(2000, 2023)
ax1.set_ylim(0, 7000000)
ax1.set_xticks([2000, 2005, 2010, 2015, 2023])
ax1.get_yaxis().set_visible(False) # Hide y-axis line/labels to match UI

# Titles and borders
ax1.set_title('Evolution of crude oil\nproduction, Nigeria', loc='left', color='#555555', weight='bold', pad=20)
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)
ax1.spines['left'].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
# CHART 2: Trade in crude oil, Nigeria (Bar Chart)
# -------------------------------------------------------------------------
fig2, ax2 = plt.subplots(figsize=(6, 5), facecolor='#F9F9F9')
ax2.set_facecolor('#F9F9F9')

# Plot the bar chart
ax2.bar(years, exports_data, color='#4A7BB0', width=0.6)

# Labels and reference line
ax2.axhline(y=6000000, color='#999999', linestyle='--', linewidth=0.8)
ax2.text(2023, 6100000, '6 000 000 TJ', ha='right', va='bottom', color='#555555', fontsize=9)
ax2.text(2000, max(exports_data) * 1.05, 'Exports', color='#333333', weight='bold', fontsize=10)

# Axis formatting
ax2.set_xlim(1999, 2024)
ax2.set_ylim(0, 7000000)
ax2.set_xticks([2000, 2005, 2010, 2015, 2023])
ax2.set_xticklabels(['20...', '2005', '2010', '2015', '20...']) # Replicating the cropped UI text look
ax2.get_yaxis().set_visible(False)

# Titles and borders
ax2.set_title('Trade in crude oil, Nigeria', loc='left', color='#555555', weight='bold', pad=20)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.spines['left'].set_visible(False)

plt.tight_layout()
plt.show()

```
