```python
# 1. Transcribe the data from the image
years = np.array([
    2000, 2001, 2002, 2003, 2004, 2005, 2006, 2007, 2008, 2009, 
    2010, 2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018, 2019, 
    2020, 2021, 2022, 2023
])

crude_oil_tj = np.array([
    4923513, 5160417, 4429358, 5041538, 5449430, 5499316, 5228556, 4880595, 
    4668570, 4678701, 5407965, 5239072, 5184954, 4693492, 4691141, 4535371, 
    3917677, 3979130, 4084111, 4295487, 3307055, 2944359, 2555025, 2737203
])
```

```python
# 2. Polynomial Least-Squares Regression (e.g., Degree 3 / Cubic Regression)
# You can change the degree integer to 2, 4, etc., depending on your needs.
poly_degree = 3 
poly_coefficients = np.polyfit(years, crude_oil_tj, poly_degree)
poly_function = np.poly1d(poly_coefficients)
```

```python

# 3. Cubic Spline Interpolation
# This forces the curve exactly through every single data point
cubic_spline = CubicSpline(years, crude_oil_tj)
```

```python
# 4. Generate smooth data points for plotting curves
years_smooth = np.linspace(years.min(), years.max(), 500)
poly_fit_vals = poly_function(years_smooth)
spline_fit_vals = cubic_spline(years_smooth)
```

```python
# --- Print Regression Results ---
print("--- Polynomial Regression Coefficients (Highest degree first) ---")
print(poly_coefficients)
print(f"\nResulting Equation (Degree {poly_degree}):")
print(poly_function)
```
