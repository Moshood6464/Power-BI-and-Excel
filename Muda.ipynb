```python
import matplotlib.pyplot as plt
import pandas as pd

# Sample dataframe setup based on your data
data = {
    'Year': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
    'Production': [2100000, 2050000, 1980000, 1920000, 1880000, 1800000, 1700000, 1500000, 1200000, 1562745],
    'Consumption': [380000, 390000, 400000, 410000, 430000, 450000, 460000, 470000, 480000, 492939],
    'Exports': [1720000, 1660000, 1580000, 1510000, 1450000, 1350000, 1240000, 1030000, 720000, 1069806]
}
df = pd.DataFrame(data)

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(df['Year'], df['Production'], marker='o', label='Production', linewidth=2)
plt.plot(df['Year'], df['Consumption'], marker='s', label='Consumption', linewidth=2)
plt.plot(df['Year'], df['Exports'], marker='^', label='Exports', linewidth=2)

plt.title('Nigeria Oil Trends: Production, Consumption & Exports (2015-2024)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Barrels per Day', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.xticks(df['Year'])
plt.tight_layout()
plt.show()

```

```python
# Breakdown of production
plt.figure(figsize=(10, 6))
plt.bar(df['Year'], df['Consumption'], label='Domestic Consumption', color='#FF9999')
plt.bar(df['Year'], df['Exports'], bottom=df['Consumption'], label='Exports', color='#66B3FF')

plt.title('Annual Oil Production Breakdown in Nigeria', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Barrels per Day', fontsize=12)
plt.xticks(df['Year'])
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Production trends vs proof reserve
df['Reserves'] = [37, 37, 37.1, 37.1, 37, 37, 37.1, 37.2, 37.3, 37.5]

fig, ax1 = plt.subplots(figsize=(10, 6))

# Primary Y-axis for Production
color = 'tab:blue'
ax1.set_xlabel('Year')
ax1.set_ylabel('Production (Barrels/Day)', color=color)
ax1.plot(df['Year'], df['Production'], color=color, marker='o', linewidth=2)
ax1.tick_params(axis='y', labelcolor=color)

# Secondary Y-axis for Reserves
ax2 = ax1.twinx()  
color = 'tab:green'
ax2.set_ylabel('Reserves (Billion Barrels)', color=color)
ax2.plot(df['Year'], df['Reserves'], color=color, marker='x', linestyle='--', linewidth=2)
ax2.tick_params(axis='y', labelcolor=color)

plt.title('Nigeria Oil Production vs. Proven Reserves', fontsize=14)
plt.xticks(df['Year'])
fig.tight_layout()
plt.show()
```
